```python
%matplotlib inline
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
# Função auxiliar para carregar e exibir a imagem
def show_image(ax, path, title=""):
    if os.path.exists(path):
        img = mpimg.imread(path)
        ax.imshow(img)
        if title:
            ax.set_title(title, fontsize=14, pad=10)
    else:
        ax.text(0.5, 0.5, f"Imagem não\nencontrada", 
                ha='center', va='center', fontsize=12, color='red')
    ax.axis('off')
```

```python
# Defina os caminhos das imagens (ajuste se seu notebook estiver em outra pasta)
original_path = "../data/cliff-walk-at-pourville.jpg"

reconstructed_paths = [
    "cliff-walk-at-pourville/reconstructed_4x4.png",
    "cliff-walk-at-pourville/reconstructed_8x8.png",
    "cliff-walk-at-pourville/reconstructed_16x16.png"
]

palette_paths = [
    "cliff-walk-at-pourville/palette_4x4.png",
    "cliff-walk-at-pourville/palette_8x8.png",
    "cliff-walk-at-pourville/palette_16x16.png"
]

labels = ["4x4", "8x8", "16x16"]

# Cria a figura (2 linhas x 4 colunas)
# gridspec_kw define a proporção (imagens maiores em cima, paletas menores embaixo)
fig, axes = plt.subplots(2, 4, figsize=(20, 8), gridspec_kw={'height_ratios': [2, 1]})

# 1. Coluna 0: Imagem Original
show_image(axes[0, 0], original_path, "Original")
axes[1, 0].axis('off') # Remove o eixo vazio na linha de baixo

# 2. Colunas 1, 2 e 3: Imagens reconstruídas e paletas
for i in range(3):
    col = i + 1
    show_image(axes[0, col], reconstructed_paths[i], f"Grid {labels[i]}")
    show_image(axes[1, col], palette_paths[i], f"Palette {labels[i]}")

# Ajusta o layout para ficar mais limpo
plt.tight_layout()

# Exibe o resultado na saída da célula
plt.show()
```

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_selected_metrics():
    # Carregar os dados do arquivo
    df = pd.read_csv("metrics.csv")
    
    # Identificar as imagens únicas mantendo a ordem original do CSV
    imagens_unicas = df['image'].unique()
    
    # Criar um dicionário mapeando o nome original para "Obra 1", "Obra 2", etc.
    mapeamento_obras = {imagem: f"Obra {i+1}" for i, imagem in enumerate(imagens_unicas)}
    
    # Aplicar o mapeamento à coluna 'image'
    df['image'] = df['image'].map(mapeamento_obras)
    
    # Lista com apenas as métricas solicitadas
    metricas = ['QE', 'MSE', 'PSNR']
    
    # Configurar o estilo visual
    sns.set_theme(style="whitegrid")
    
    # Criar uma figura com 1 linha e 3 colunas
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    for i, metrica in enumerate(metricas):
        ax = axes[i]
        
        # Criar o gráfico de linhas
        sns.lineplot(
            data=df, 
            x='side', 
            y=metrica, 
            hue='image', 
            marker='o', 
            markersize=8,
            linewidth=2,
            ax=ax
        )
        
        # Configurar títulos e eixos
        ax.set_title(f'Evolução de {metrica}', fontsize=14, fontweight='bold')
        ax.set_xlabel('Tamanho do Grid', fontsize=12)
        ax.set_ylabel(f'Valor ({metrica})', fontsize=12)
        
        # Definir primeiro as posições numéricas e depois os rótulos de texto
        ax.set_xticks([4, 8, 16])
        ax.set_xticklabels(["4x4", "8x8", "16x16"])
        
        # Deixar a legenda apenas no último gráfico
        if i == 2:
            ax.legend(title='Obras', bbox_to_anchor=(1.05, 1), loc='upper left')
        else:
            if ax.get_legend():
                ax.get_legend().remove()

    # Ajustar o espaçamento
    plt.tight_layout()
    plt.show()

# Executar a função
plot_selected_metrics()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from pathlib import Path

def plot_todas_obras(data_dir="../data"):
    # Definir o diretório
    path_obj = Path(data_dir)
    
    # Procurar por extensões de imagem comuns
    image_paths = []
    for ext in ["*.jpg", "*.jpeg", "*.png"]:
        image_paths.extend(path_obj.glob(ext))
        
    # Ordenar alfabeticamente para manter a ordem consistente
    image_paths = sorted(image_paths)
    n_images = len(image_paths)
    
    if n_images == 0:
        print(f"Nenhuma imagem encontrada no diretório: {data_dir}")
        return
        
    print(f"Foram encontradas {n_images} imagens.")
        
    # Criar a figura (1 linha, N colunas)
    # A largura da figura (figsize) aumenta 5 unidades por cada imagem encontrada
    fig, axes = plt.subplots(1, n_images, figsize=(5 * n_images, 5))
    
    # Se houver apenas 1 imagem, o matplotlib não devolve uma matriz de eixos, 
    # por isso transformamos numa lista para o ciclo "for" funcionar
    if n_images == 1:
        axes = [axes]
        
    for i, img_path in enumerate(image_paths):
        ax = axes[i]
        try:
            # Carregar e exibir a imagem
            img = mpimg.imread(str(img_path))
            ax.imshow(img)
            
            # Definir o título como "Obra 1", "Obra 2", etc.
            ax.set_title(f"Obra {i+1}", fontsize=16, fontweight='bold', pad=10)
        except Exception as e:
            ax.text(0.5, 0.5, f"Erro ao ler:\n{img_path.name}", 
                    ha='center', va='center', fontsize=12, color='red')
            
        # Ocultar os eixos (linhas e números)
        ax.axis('off')
        
    # Ajustar o layout
    plt.tight_layout()
    plt.show()

# Executar a função
plot_todas_obras()
```
